```sql
%sql
CREATE SCHEMA IF NOT EXISTS azure_databricks_dev.bronze;

-- Change this base path if you place the CSV files in a different Volume.
-- Expected layout:
-- /Volumes/azure_databricks_dev/bronze/raw/events.csv
-- /Volumes/azure_databricks_dev/bronze/raw/orders.csv
-- /Volumes/azure_databricks_dev/bronze/raw/order_items.csv
-- /Volumes/azure_databricks_dev/bronze/raw/products.csv
-- /Volumes/azure_databricks_dev/bronze/raw/reviews.csv
-- /Volumes/azure_databricks_dev/bronze/raw/users.csv
```

```sql
%sql
CREATE OR REPLACE TABLE azure_databricks_dev.bronze.events_raw
USING DELTA
AS
SELECT *
FROM read_files(
  '/Volumes/azure_databricks_dev/bronze/raw/events.csv',
  format => 'csv',
  header => true,
  inferSchema => true
);
```

```sql
-- Basic Bronze row-count check
SELECT 'events' AS table_name, COUNT(*) AS row_count FROM azure_databricks_dev.bronze.events_raw
UNION ALL
SELECT 'orders', COUNT(*) FROM azure_databricks_dev.bronze.orders_raw
UNION ALL
SELECT 'order_items', COUNT(*) FROM azure_databricks_dev.bronze.order_items_raw
UNION ALL
SELECT 'products', COUNT(*) FROM azure_databricks_dev.bronze.products_raw
UNION ALL
SELECT 'reviews', COUNT(*) FROM azure_databricks_dev.bronze.reviews_raw
UNION ALL
SELECT 'users', COUNT(*) FROM azure_databricks_dev.bronze.users_raw;
```
